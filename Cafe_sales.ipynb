### Step 8: Handle Missing Values

```python
df.isna().sum()
```

```python
(df.isna().sum() / len(df)) * 100
```

```python
df["quantity"].describe()
```

```python
df[["quantity", "price_per_unit", "total_spent"]].isna().any(axis=1).sum()
```

```python
df[["quantity", "price_per_unit", "total_spent"]].isna().value_counts()
```

```python
df.loc[
    df["total_spent"].isna() &
    df["quantity"].notna() &
    df["price_per_unit"].notna(),
    "total_spent"
] = (
    df["quantity"] * df["price_per_unit"]
)
```

```python
df["total_spent"].isna().sum()
```

```python
df.loc[
    df["price_per_unit"].isna() &
    df["quantity"].notna() &
    df["total_spent"].notna(),
    "price_per_unit"
] = (
    df["total_spent"] / df["quantity"]
)
```

```python
df["price_per_unit"].isna().sum()
```

```python
df.loc[
    df["quantity"].isna() &
    df["price_per_unit"].notna() &
    df["total_spent"].notna(),
    "quantity"
] = (
    df["total_spent"] / df["price_per_unit"]
)
```
